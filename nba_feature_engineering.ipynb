# NBA Feature Engineering Evaluation


Objective: predict target_5yrs and engineer model-ready features.

```
import pandas as pd
df=pd.read_csv("nba_player_dataset.csv")
df.head()
```

Target variable: `target_5yrs` (dependent variable). Removed name and index columns to reduce leakage/noise.

```
# Drop noisy columns
df=df.drop(columns=["Unnamed: 0","name"])
```

Correlation review showed strong multicollinearity (FGM/FGA, 3PA, FTA, REB). Selected representative variables.

```
df["points_per_minute"]=df["pts"]/df["min"]
df["efficiency_rating"]=(df["pts"]+df["reb"]+df["ast"]+df["stl"]+df["blk"]-df["tov"])
```

Missing values handled using median imputation. Dataset becomes ML-ready.
